FUNCTION TO ADD EXPENSE

```python
def add_expense(expenses, payer, amount, participants):

    share = amount / len(participants)

    for person in participants:
        expenses[person] -= share

    expenses[payer] += amount
```

FUNCTION TO SHOW EXPENSE

```python
def show_expenses(expenses):

    print("\nExpense Summary")

    for person, amount in expenses.items():
        print(person, ":", round(amount, 2))
```

FUNCTION FOR SETTLEMENT

```python
def settlement(expenses):

    print("\nFinal Settlement")

    for person, amount in expenses.items():

        if amount > 0:
            print(person, "should receive Rs.", round(amount, 2))

        elif amount < 0:
            print(person, "owes Rs.", round(abs(amount), 2))

        else:
            print(person, "is settled")
```

MAIN PROGRAM


```python
friends = input(
    "Enter names of friends separated by commas: "
).split(",")

friends = [name.strip() for name in friends]

expenses = {}

for friend in friends:
    expenses[friend] = 0

while True:

    payer = input(
        "Enter payer name (or done): "
    )

    if payer.lower() == "done":
        break

    amount = float(input("Enter amount paid: "))

    participants = input(
        "Enter participants separated by commas: "
    ).split(",")

    participants = [name.strip() for name in participants]

    add_expense(
        expenses,
        payer,
        amount,
        participants
    )

show_expenses(expenses)

settlement(expenses)
```
